## SALES DATA ANALYSIS

## Business Problem

The retail company is preparing its monthly sales performance report, but the raw sales data contains missing values, duplicate records, inconsistent formats, incorrect data types, invalid numerical values, and potentially incorrect Sales and Profit values. If these issues are not resolved before analysis, management may make decisions using inaccurate revenue, profitability, product, regional, and salesperson information. Your task is to act as a Data Analyst and convert the raw sales file into a reliable, consistent, and analysis-ready dataset that can be safely used for reporting and future EDA.

## 4. Data Type & Format Correction

```python
print(df.dtypes)
```

1. Convert Quantity to numeric

```python
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
```

2. Convert Unit_Price to numeric

```python
df['Unit_Price'] = df['Unit_Price'].astype(str).str.extract(r'(\d+\.?\d*)')[0].astype(float)
```

3. Convert Discount, Sales and Profit

```python
df[['Discount','Sales','Profit']] = df[['Discount','Sales','Profit']].apply(pd.to_numeric, errors='coerce')
```

4. Convert Order_Date

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
```
